```python
# ==========================================
#Instruction-Tuning Dataset Preparation
# ==========================================
from flask import json
import pandas as pd
import gzip
import os
import random
import re
import nltk
from nltk.corpus import words
import torch
from torch.utils.data import Dataset, DataLoader
```

```python
# -------------------------------
# 1. Setup
# -------------------------------
nltk.download('words')
english_words = set(words.words())
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Running on:", device)
```

```python
# -------------------------------
# 2. Read JSONL Files
# -------------------------------
def parse(path):
    with gzip.open(path, 'rb') as g:
        for line in g:
            yield json.loads(line)

def getDF(path):
    data = {}
    for i, d in enumerate(parse(path)):
        data[i] = d
    return pd.DataFrame.from_dict(data, orient='index')
```

```python
# Update file paths as needed
dfd = getDF(r"C:\Users\792417\OneDrive - hull.ac.uk\Workspace\Lab 4 - Fine Tuning LoRA\Lab-4-Fine-Tuning-with-LoRA\All_Beauty.jsonl.gz") # review data
dfm = getDF( r"C:\Users\792417\OneDrive - hull.ac.uk\Workspace\Lab 4 - Fine Tuning LoRA\Lab-4-Fine-Tuning-with-LoRA\meta_All_Beauty.jsonl.gz") # metadata
```

```python
# -------------------------------
# 3. Drop Unnecessary Columns
# -------------------------------
# Drop columns only if they exist (for safety)
dfd.drop(columns=[c for c in [
'rating', 'title', 'text', 'images', 'timestamp',
'helpful_vote', 'verified_purchase'
] if c in dfd.columns], inplace=True, errors='ignore')

dfm.drop(columns=[c for c in [
'main_category', 'average_rating', 'rating_number', 'features',
'description', 'price', 'images', 'videos', 'store',
'categories', 'details', 'bought_together'
] if c in dfm.columns], inplace=True, errors='ignore')
```

```python
# -------------------------------
# 4. Filter Active Users
# -------------------------------
reviewer_counts = dfd['user_id'].value_counts()
reviewer_ids_at_least_5 = reviewer_counts[reviewer_counts >= 5].index
dfd = dfd[dfd['user_id'].isin(reviewer_ids_at_least_5)]
dfd.reset_index(inplace=True, drop=True)
```

```python
# -------------------------------
# 5. Merge Data on 'parent_asin'
# -------------------------------
df = pd.merge(dfd, dfm, on='parent_asin', how='inner')
```

```python
# -------------------------------
# 6. Clean Product Titles
# -------------------------------
def clean_text(text):
    if not isinstance(text, str):
        text = str(text)
    text = re.sub(r'[^\w\s\.,\?]|(?<=\s)\w(?=\s)', ' ', text) # remove special chars
    text = re.sub(r'\b\w*[\d_]+\w*\b', '', text) # remove digits/underscores
    text = re.sub(r'\bmisspelled\b|\bmeaningless\b|\bwordss\b|\bsymbols\b', ' ', text,
    flags=re.IGNORECASE)
    text = ' '.join(word for word in text.split() if word.lower() in english_words) # keep English words only
    text = re.sub(r'\b\w\b', ' ', text) # remove single chars
    text = re.sub(r'\s+', ' ', text) # normalize spaces
    return text.strip()


if 'title' in df.columns:
    df['title'] = df['title'].apply(clean_text)
else:
    raise KeyError("No title column found after merging. Please check dfm columns.")
```

```python
# -------------------------------
# 7. Group by User and Build Histories
# -------------------------------
grouped_df = df.groupby('user_id').apply(lambda x: {
    'data': {
        'total_asin': len(x),
        **{
            str(i + 1): x.iloc[i, 1:].to_dict() # store each item as dict
            for i in range(len(x))
        }
    }
}).reset_index()

grouped_df.columns = ['ID', 'info']
```

```python
# -------------------------------
# 8. Extract First 5 Item Titles
# -------------------------------
for i in range(1, 6):
    grouped_df[f'item_{i}_title'] = grouped_df['info'].apply(
        lambda x: x['data'].get(str(i), {}).get('title', '')
)
```

```python
# -------------------------------
# 9. Create Prompts Sample and Outputs
# -------------------------------
grouped_df['prompts'] = (
    "Reviewer has bought " +
    grouped_df['item_1_title'] + ", " +
    grouped_df['item_2_title'] + ", " +
    grouped_df['item_3_title'] + ", " +
    grouped_df['item_4_title'] +
    ". Considering their purchase history, which product will they buy next from the below options?"
)
grouped_df['outputs'] = grouped_df['item_5_title']
```

```python
# -------------------------------
# 10. Add Negative Samples
# -------------------------------
def get_titles(asins):
    titles = []
    for asin in asins:
        t = dfm.loc[dfm['parent_asin'] == asin, 'title'].iloc[0]
        titles.append(t)
    return titles

def create_additional_columns(row):
    total_asin = row['info']['data']['total_asin']
    used_asins = [row['info']['data'][str(i)]['parent_asin'] for i in range(1, total_asin + 1)]
    available_asins = dfm[~dfm['parent_asin'].isin(used_asins)]['parent_asin'].tolist()
    random.shuffle(available_asins)
    random_items = random.sample(available_asins, 3)
    return pd.Series(get_titles(random_items), index=['item_1_out', 'item_2_out', 'item_3_out'])

```

```python
grouped_df[['item_1_out', 'item_2_out', 'item_3_out']] = grouped_df.apply(create_additional_columns, axis=1)
```

```python
# -------------------------------
# 11. Build Final Prompt Options
# -------------------------------
def generate_prompt(row):
    items = [row['outputs'], row['item_1_out'], row['item_2_out'], row['item_3_out']]
    random.shuffle(items)
    return f" A) {items[0]}, B) {items[1]}, C) {items[2]}, D) {items[3]}."

grouped_df['prompts_part2'] = grouped_df.apply(generate_prompt, axis=1)
grouped_df['prompts'] = grouped_df['prompts'].str.cat(grouped_df['prompts_part2'], sep=' ')
```

```python
# -------------------------------
# 12. Save Final Dataset
# -------------------------------
LLM_data = grouped_df[['prompts', 'outputs']]
print(LLM_data.head())
LLM_data.to_csv(r"C:\Users\792417\OneDrive - hull.ac.uk\Workspace\Lab 4 - Fine Tuning LoRA\Lab-4-Fine-Tuning-with-LoRA\final_dataset_sft.csv", index=False)
print(" Final dataset saved successfully!")
```

```python
# LoRA Fine-tuning(Jupyter version)
# =====================================================================
# - Works on both CPU and GPU
# - Automatically disables bitsandbytes if CUDA unavailable
# - Dataset must have: columns ['prompts', 'outputs']

import os, json, torch, random
import pandas as pd
from dataclasses import dataclass
from typing import List, Dict, Optional
from transformers import (
AutoTokenizer, AutoModelForSeq2SeqLM, 
Trainer, TrainingArguments, set_seed,
)
from peft import LoraConfig, get_peft_model
```

```python
# ---------------- CONFIG ----------------
data_path = "C:\\Users\\792417\\OneDrive - hull.ac.uk\\Workspace\\Lab 4 - Fine Tuning LoRA\\Lab-4-Fine-Tuning-with-LoRA\\final_dataset_sft.csv" # generated in excerise 1
output_dir = "C:\\Users\\792417\\OneDrive - hull.ac.uk\\Workspace\\Lab 4 - Fine Tuning LoRA\\Lab-4-Fine-Tuning-with-LoRA" # save dir
model_name = "google-t5/t5-small" # Using T5 model

num_train_epochs = 5
learning_rate = 3e-4
batch_size = 8
max_seq_len = 1024
eval_top_k = 5
eval_subset = 50
eval_max_new_tokens = 128
seed = 42
set_seed(seed)
```

```python
# ---------------- ENV / DEVICE ----------------
device = "cuda" if torch.cuda.is_available() else "cpu"
load_in_4bit = torch.cuda.is_available() # only if GPU present
print(f"Device: {device} | 4-bit: {load_in_4bit}")
```

```python
# ---------------- LOAD DATA ----------------
df = pd.read_csv(data_path).dropna(subset=["prompts", "outputs"])
df["prompts"], df["outputs"] = df["prompts"].astype(str), df["outputs"].astype(str)
```

```python
# Split data into train, validation, and test sets (80/10/10 split)
train_split = int(0.8 * len(df))
val_split = int(0.9 * len(df)) # Remaining 20% for val and test

train_df = df.iloc[:train_split]
val_df = df.iloc[train_split:val_split]
test_df = df.iloc[val_split:]
print(f"Train set size: {len(train_df)}")
print(f"Validation set size: {len(val_df)}")
print(f"Test set size: {len(test_df)}")
```

```python
# ---------------- TOKENIZER + MODEL ----------------
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
# T5 does not have a pad_token by default, using eos_token as pad_token
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

print("Loading model...")
model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name,
    trust_remote_code=True,
    device_map="auto" if torch.cuda.is_available() else None,
    torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
    #load_in_4bit=False,
 )
model.to(device)
```

```python
# ---------------- LoRA ATTACH ----------------
# LoRA target modules for T5 models
# Common target modules for T5 are 'q', 'v'
cand_sets = [
    ["q", "v"], # Common T5 target modules
    ["q_proj","k_proj","v_proj","o_proj","gate_proj","up_proj","down_proj"],
    ["W_pack","o_proj"], ["c_attn","c_proj","c_fc"], ["qkv_proj","out_proj","fc_in","fc_out"],
]
found = set(name for name, _ in model.named_modules())
chosen = next((c for c in cand_sets if any(any(x in n for x in c) for n in found)), ["q",
    "v"]) # Default to q, v for T5
print("LoRA target modules:", chosen)


lora_cfg = LoraConfig(r=16, lora_alpha=32, lora_dropout=0.05,
                    task_type="SEQ_2_SEQ_LM", # Changed task_type
                    target_modules=chosen)
model = get_peft_model(model, lora_cfg)
model.config.use_cache = False # Disable cache for training
model.print_trainable_parameters()
system_prompt = "You are a concise, helpful assistant."
```

```python
# ---------------- PREP DATA ----------------
def normalize_text(s: str): return " ".join(s.lower().strip().split()) if s else ""

# Modified build_text for Seq2Seq models (T5)
def build_text(prompt, output):
    # For T5, prompt goes into the encoder, output is the target for the decoder
    # The tokenizer handles the correct formatting (e.g., adding prefix)
    return prompt, output

@dataclass
class SFTItem:
    input_ids: List[int]; attention_mask: List[int]; labels: List[int] # Reordered for clarity

def make_item(prompt, output):
    # Tokenize prompt for input
    input_tok = tokenizer(prompt, truncation=True, max_length=max_seq_len,
        return_tensors="pt")
    input_ids = input_tok["input_ids"].squeeze().tolist()
    attention_mask = input_tok["attention_mask"].squeeze().tolist()
    
# Tokenize output for labels (decoder target)
    # Use as_target_tokenizer to tokenize for the decoder side (as_target_tokenizer is not there anymore??)
    label_tok = tokenizer(
        text_target=output, 
        truncation=True, 
        max_length=max_seq_len,
        return_tensors="pt"
        )
    labels = label_tok["input_ids"].squeeze().tolist()

    return SFTItem(input_ids, attention_mask, labels)

class DS(torch.utils.data.Dataset):
    def __init__(self, df): self.items=[make_item(p,o) for p,o in zip(df.prompts,df.outputs)]
    def __len__(self): return len(self.items)
    def __getitem__(self,i): it=self.items[i]; return {
        "input_ids":torch.tensor(it.input_ids),
        "attention_mask":torch.tensor(it.attention_mask),
        "labels":torch.tensor(it.labels),
    }

train_ds, val_ds,test_ds = DS(train_df), DS(val_df),DS(test_df)
```

```python
# ---------------- PAD COLLATOR ----------------
"""
A data collator dynamically pads examples within each batch so that
they all share the same sequence length.
This is essential for efficient GPU batching, since sequences often
have variable lengths.
The collator ensures that all input tensors in a batch align correctly
and that loss computation ignores padded positions.
"""
@dataclass
class DataCollatorForSeq2SeqWithPadding: # Changed collator name
    tokenizer: any
    pad_to_multiple_of: Optional[int] = 8

    def __call__(self, features):
        # convert tensors lists (Trainer may pass tensors)
        features = [
            {
            k: (v.tolist() if isinstance(v, torch.Tensor) else v)
            for k, v in f.items()
            }
            for f in features
    ]

        # Get padding token IDs for input and labels
        input_pad_id = self.tokenizer.pad_token_id or self.tokenizer.eos_token_id
        # For labels, T5 uses -100 for padding in the loss calculation
        label_pad_id = -100

        max_input_len = max(len(f["input_ids"]) for f in features)
        max_label_len = max(len(f["labels"]) for f in features)

        # Round lengths for performance
        if self.pad_to_multiple_of:
            m = self.pad_to_multiple_of
            max_input_len = (max_input_len + m - 1) // m * m
            max_label_len = (max_label_len + m - 1) // m * m


        def pad(seq, max_len, pad_value):
            seq = list(seq)
            if len(seq) < max_len:
                seq = seq + [pad_value] * (max_len - len(seq))
            return seq[:max_len]

        input_ids = [torch.tensor(pad(f["input_ids"], max_input_len, input_pad_id),
            dtype=torch.long) for f in features]
        attention_mask = [torch.tensor(pad(f["attention_mask"], max_input_len, 0),
            dtype=torch.long) for f in features]
        labels = [torch.tensor(pad(f["labels"], max_label_len, label_pad_id),
            dtype=torch.long) for f in features] # Corrected padding for labels

        return {
            "input_ids": torch.stack(input_ids),
            "attention_mask": torch.stack(attention_mask),
            "labels": torch.stack(labels),
        }

# Instantiate correctly with Seq2Seq collator
collator = DataCollatorForSeq2SeqWithPadding(tokenizer=tokenizer)
```

```python
# ---------------- TRAIN ----------------
args = TrainingArguments(
    output_dir=output_dir,
    num_train_epochs=num_train_epochs,
    per_device_train_batch_size=batch_size,
    gradient_accumulation_steps=4,
    learning_rate=learning_rate,
    logging_steps=10,
    save_strategy="epoch",
    bf16=torch.cuda.is_available() and torch.cuda.is_bf16_supported(),
    report_to="none",
    # group_by_length=True, # May not be directly supported or as effective for Seq2Seq
)

trainer = Trainer(model=model, args=args, train_dataset=train_ds,
                  #its processing_class=tokenizer as huggingface renamed it from tokenizer=tokenizer
                eval_dataset=val_ds, data_collator=collator, processing_class=tokenizer) # Pass tokenizer to Trainer
trainer.train()

model.save_pretrained(f"{output_dir}/lora_adapter_ft")
tokenizer.save_pretrained(output_dir)
```

```python
# ======================================================
# EVALUATION Hit@1/3/5 & NDCG@1/3/5 + show generated text
# ======================================================
import os, json, torch, numpy as np, pandas as pd
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from peft import PeftModel

# ---------------- CONFIG ----------------
model_name = "google-t5/t5-small"
adapter_path = r"C:\Users\792417\OneDrive - hull.ac.uk\Workspace\Lab 4 - Fine Tuning LoRA\Lab-4-Fine-Tuning-with-LoRA\lora_adapter_ft"
eval_max_new_tokens = 128
system_prompt = "You are a concise, helpful assistant."
device = "cuda" if torch.cuda.is_available() else "cpu"
#
# We'll compute @1, @3, @5
K_values = [1, 3, 5]
max_K = max(K_values)
```

```python
# ---------------- LOAD MODEL ----------------
print("Loading base model...")
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
base_model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name,
    device_map="auto" if torch.cuda.is_available() else None,
    torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
)
print("Loading LoRA adapter...")
model = PeftModel.from_pretrained(base_model, adapter_path)
model.to(device)
model.eval()
```

```python
# ---------------- NORMALIZATION ----------------
def normalize_text(s: str):
    return " ".join(s.lower().strip().split()) if s else ""
```

```python
# ---------------- METRICS INIT ----------------
hit_counts = {k: 0 for k in K_values}
ndcg_scores = {k: [] for k in K_values}
total = len(test_df)

# The 'test_df' was created in Exercise 2 (last 10% of dataset)
print(f"\nStarting evaluation on {total} samples...\n")
for i, row in test_df.iterrows():
    prompt, gold = row.prompts, row.outputs

    # Prepare input
    gen_text = f"<|system|>\n{system_prompt}\n<|user|>\n{prompt}\n<|assistant|>\n"
    inputs = tokenizer(gen_text, return_tensors="pt").to(device)

    # Generate up to max_K sequences
    with torch.no_grad():
        outs = model.generate(
    **inputs,
    max_new_tokens=eval_max_new_tokens,
    num_beams=max_K,
    num_return_sequences=max_K,
    pad_token_id=tokenizer.pad_token_id,
    )
    seqs = [s.strip() for s in tokenizer.batch_decode(outs, skip_special_tokens=True)]

    # ---- Metrics ----
    normalized_gold = normalize_text(gold)
    normalized_preds = [normalize_text(s) for s in seqs]

    # Compute Hit@K and NDCG@K for each K value
    for K in K_values:
        topK = normalized_preds[:K]
        if normalized_gold in topK:
            hit_counts[K] += 1

            # Find the rank of the gold item (1-based)
            rank = topK.index(normalized_gold) + 1
            ndcg_scores[K].append(1 / np.log2(rank + 1))
        else:
            ndcg_scores[K].append(0.0)

# ---- Display sample ----
print(f"\nSample {i+1}/{total}")
print(f"Prompt: {prompt}")
print(f"Gold Output: {gold}")
for j, seq in enumerate(seqs):
    print(f" Gen[{j+1}]: {seq}")
print("-" * 80)
if (i + 1) % 10 == 0:
    print(f"Evaluated {i + 1}/{total}")
```

```python
# ---------------- RESULTS ----------------
results = {}
for K in K_values:
    results[f"Hit@{K}"] = hit_counts[K] / total
    results[f"NDCG@{K}"] = float(np.mean(ndcg_scores[K]))

print("\n=== FINAL RESULTS ===")
for K in K_values:
    print(f"Hit@{K}: {results[f'Hit@{K}']:.3f} | NDCG@{K}: {results[f'NDCG@{K}']:.3f}")
```

```python
# ---------------- SAVE ----------------
os.makedirs(adapter_path, exist_ok=True)
result_path = os.path.join(adapter_path, "eval_results.json")
with open(result_path, "w") as f:
    json.dump(results, f, indent=2)
print(f"\ n Results saved to {result_path}")
```
